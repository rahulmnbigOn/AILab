```python
import heapq

# Initial and goal states
initial = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

goal = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)


# ---------------------------------------------------------
# Generate possible next states
# ---------------------------------------------------------
def get_neighbors(state):
    neighbors = []

    # Find blank (0)
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                x, y = i, j

    # Up, Down, Left, Right
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dx, dy in moves:
        nx, ny = x + dx, y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [list(row) for row in state]

            # Swap blank with adjacent tile
            new_state[x][y], new_state[nx][ny] = \
                new_state[nx][ny], new_state[x][y]

            neighbors.append(tuple(tuple(row) for row in new_state))

    return neighbors


# ---------------------------------------------------------
# Heuristic 1: Number of misplaced tiles
# ---------------------------------------------------------
def misplaced_tiles(state):
    count = 0

    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != goal[i][j]:
                count += 1

    return count


# ---------------------------------------------------------
# Heuristic 2: Manhattan distance
# ---------------------------------------------------------
def manhattan_distance(state):
    distance = 0

    # Find goal position of every tile
    goal_position = {}

    for i in range(3):
        for j in range(3):
            goal_position[goal[i][j]] = (i, j)

    for i in range(3):
        for j in range(3):
            tile = state[i][j]

            if tile != 0:
                gi, gj = goal_position[tile]
                distance += abs(i - gi) + abs(j - gj)

    return distance


# ---------------------------------------------------------
# A* Search
# ---------------------------------------------------------
def a_star(initial, heuristic):
    # Priority queue:
    # (f, g, state, path)
    pq = []

    h = heuristic(initial)

    heapq.heappush(
        pq,
        (h, 0, initial, [initial])
    )

    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        # Goal reached
        if state == goal:
            return path, g

        # Generate children
        for next_state in get_neighbors(state):

            if next_state not in visited:
                new_g = g + 1
                new_h = heuristic(next_state)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (
                        new_f,
                        new_g,
                        next_state,
                        path + [next_state]
                    )
                )

    return None, -1


# ---------------------------------------------------------
# Print solution
# ---------------------------------------------------------
def print_solution(path, cost):
    print("\nSolution found!")
    print("Number of moves:", cost)
    print()

    for step, state in enumerate(path):
        print("Step", step)

        for row in state:
            print(*row)

        print()


# =========================================================
# CASE 1
# g(n) = depth
# h(n) = number of misplaced tiles
# =========================================================

print("========== CASE 1 ==========")
print("Heuristic: Misplaced Tiles")

path1, cost1 = a_star(initial, misplaced_tiles)

print_solution(path1, cost1)


# =========================================================
# CASE 2
# g(n) = depth
# h(n) = Manhattan distance
# =========================================================

print("========== CASE 2 ==========")
print("Heuristic: Manhattan Distance")

path2, cost2 = a_star(initial, manhattan_distance)

print_solution(path2, cost2)
```
