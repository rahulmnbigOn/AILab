```python

# Knowledge Base (Given Facts)

facts = {
    ("Eats", "Anil", "Peanuts"),
    ("Alive", "Anil")
}

# Forward Chaining
changed = True

while changed:
    changed = False

    # Rule 1: Alive(x) -> NotKilled(x)
    for f in list(facts):
        if f[0] == "Alive":
            new_fact = ("NotKilled", f[1])

            if new_fact not in facts:
                facts.add(new_fact)
                changed = True

    # Rule 2: NotKilled(x) -> Alive(x)
    for f in list(facts):
        if f[0] == "NotKilled":
            new_fact = ("Alive", f[1])

            if new_fact not in facts:
                facts.add(new_fact)
                changed = True

    # Rule 3: Eats(x,y) AND NotKilled(x) -> Food(y)
    for e in list(facts):
        if e[0] == "Eats":
            person = e[1]
            food = e[2]

            if ("NotKilled", person) in facts:
                new_fact = ("Food", food)

                if new_fact not in facts:
                    facts.add(new_fact)
                    changed = True

    # Rule 4: Food(x) -> Likes(John,x)
    for f in list(facts):
        if f[0] == "Food":
            new_fact = ("Likes", "John", f[1])

            if new_fact not in facts:
                facts.add(new_fact)
                changed = True

# Check the Goal
goal = ("Likes", "John", "Peanuts")

print("Derived Facts:")
for f in sorted(facts):
    print(f)

if goal in facts:
    print("\nGoal Proved: John likes peanuts.")
else:
    print("\nGoal Not Proved.")
```
