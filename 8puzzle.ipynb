```python
# 8-Puzzle using Depth First Search (DFS)

def print_board(state):
    """Print the puzzle in 3 x 3 format."""
    for i in range(0, 9, 3):
        row = state[i:i + 3]
        print(" ".join("_" if x == 0 else str(x) for x in row))
    print()


def get_neighbors(state):
    """
    Generate all possible states from the current state.
    Returns: (new_state, move)
    """
    neighbors = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    # Possible moves: (row change, col change, move name)
    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc

        # Check if move is inside the board
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col

            # Convert tuple to list so we can swap
            new_state = list(state)

            # Move tile into blank position
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append((tuple(new_state), move))

    return neighbors


def dfs(start, goal):
    """Solve the 8-puzzle using Depth First Search."""

    # Stack for DFS
    stack = [start]

    # Keep track of visited states
    visited = {start}

    # Store parent of each state
    parent = {start: None}

    # Store the move used to reach each state
    move_taken = {start: None}

    while stack:

        # Take the last element -> DFS
        current = stack.pop()

        # Check whether goal is reached
        if current == goal:
            break

        # Generate neighboring states
        for next_state, move in get_neighbors(current):

            if next_state not in visited:
                visited.add(next_state)

                parent[next_state] = current
                move_taken[next_state] = move

                stack.append(next_state)

    # Goal was not found
    if goal not in parent:
        return None

    # Reconstruct solution path
    path = []
    current = goal

    while current is not None:
        path.append((current, move_taken[current]))
        current = parent[current]

    # Reverse path: initial -> goal
    path.reverse()

    return path


# --------------------------------------------------
# Main program
# --------------------------------------------------

# 0 represents the blank space
initial = (
    5, 4, 0,
    6, 1, 8,
    7, 3, 2
)

goal = (
    0, 1, 2,
    3, 4, 5,
    6, 7, 8
)

# Run DFS
solution = dfs(initial, goal)

# Display result
if solution is None:
    print("No solution exists.")
else:
    print("Solution found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for step, (state, move) in enumerate(solution):
        print("Step:", step)

        if move is not None:
            print("Move:", move)

        print_board(state)
```
