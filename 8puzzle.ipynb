```python
# ============================================================
# 8-PUZZLE SOLVER USING DEPTH FIRST SEARCH (DFS)
# ============================================================

# 0 represents the blank space


# ------------------------------------------------------------
# 1. PRINT THE BOARD
# ------------------------------------------------------------

def print_board(state):
    """Print a 3x3 puzzle board."""

    for i in range(0, 9, 3):
        row = state[i:i+3]

        print(
            " ".join(
                "_" if x == 0 else str(x)
                for x in row
            )
        )

    print()


# ------------------------------------------------------------
# 2. GENERATE POSSIBLE MOVES
# ------------------------------------------------------------

def get_neighbors(state):
    """
    Generate all possible states from the current state.

    Returns:
        (new_state, move)
    """

    neighbors = []

    # Find blank position
    blank_position = state.index(0)

    row = blank_position // 3
    col = blank_position % 3

    # Possible movements of blank
    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for row_change, col_change, move_name in moves:

        new_row = row + row_change
        new_col = col + col_change

        # Make sure the new position is inside the board
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank_position = new_row * 3 + new_col

            # Convert tuple to list
            new_state = list(state)

            # Swap blank with the tile
            new_state[blank_position], new_state[new_blank_position] = \
                new_state[new_blank_position], new_state[blank_position]

            # Convert back to tuple
            new_state = tuple(new_state)

            neighbors.append((new_state, move_name))

    return neighbors


# ------------------------------------------------------------
# 3. DEPTH FIRST SEARCH
# ------------------------------------------------------------

def dfs(start, goal, max_depth=31):

    # Stack contains:
    # (state, depth)
    stack = [(start, 0)]

    # States already visited
    visited = {start}

    # Store parent of each state
    parent = {
        start: None
    }

    # Store move used to reach each state
    move_taken = {
        start: None
    }

    # Number of states explored
    nodes_explored = 0

    while stack:

        # Remove last element from stack
        current, depth = stack.pop()

        nodes_explored += 1

        # Check if goal reached
        if current == goal:

            print("Goal found!")
            print("States explored:", nodes_explored)
            print("Solution depth:", depth)

            # Reconstruct path
            path = []

            state = current

            while state is not None:

                path.append(
                    (
                        state,
                        move_taken[state]
                    )
                )

                state = parent[state]

            # Reverse path
            path.reverse()

            return path

        # Don't go beyond depth limit
        if depth >= max_depth:
            continue

        # Generate next states
        for next_state, move in get_neighbors(current):

            if next_state not in visited:

                visited.add(next_state)

                parent[next_state] = current

                move_taken[next_state] = move

                stack.append(
                    (
                        next_state,
                        depth + 1
                    )
                )

    print("No solution found within depth limit.")

    print("States explored:", nodes_explored)

    return None


# ------------------------------------------------------------
# 4. INITIAL AND GOAL STATES
# ------------------------------------------------------------

# Initial state from your question:
#
# 5 4 _
# 6 1 8
# 7 3 2

initial_state = (
    5, 4, 0,
    6, 1, 8,
    7, 3, 2
)


# Goal state:
#
# _ 1 2
# 3 4 5
# 6 7 8

goal_state = (
    0, 1, 2,
    3, 4, 5,
    6, 7, 8
)


# ------------------------------------------------------------
# 5. DISPLAY INITIAL STATE
# ------------------------------------------------------------

print("=" * 40)
print("8-PUZZLE USING DEPTH FIRST SEARCH")
print("=" * 40)

print("\nINITIAL STATE:")
print_board(initial_state)

print("GOAL STATE:")
print_board(goal_state)


# ------------------------------------------------------------
# 6. RUN DFS
# ------------------------------------------------------------

print("Searching...\n")

solution = dfs(
    initial_state,
    goal_state,
    max_depth=31
)


# ------------------------------------------------------------
# 7. DISPLAY SOLUTION
# ------------------------------------------------------------

if solution is not None:

    print("\n")
    print("=" * 40)
    print("SOLUTION PATH")
    print("=" * 40)

    print("\nTotal moves:", len(solution) - 1)

    print("\n")

    for step, (state, move) in enumerate(solution):

        print("Step:", step)

        if move is not None:
            print("Move:", move)

        print_board(state)


else:

    print("\nNo solution was found.")
```

```python
# ============================================================
# 8-PUZZLE USING ITERATIVE DEEPENING SEARCH (IDS)
# ============================================================

# 0 represents the blank space


# ------------------------------------------------------------
# 1. PRINT BOARD
# ------------------------------------------------------------

def print_board(state):
    for i in range(0, 9, 3):
        row = state[i:i + 3]

        print(
            " ".join(
                "_" if x == 0 else str(x)
                for x in row
            )
        )

    print()


# ------------------------------------------------------------
# 2. CHECK WHETHER PUZZLE IS SOLVABLE
# ------------------------------------------------------------

def is_solvable(state):

    # Remove blank
    numbers = [x for x in state if x != 0]

    inversions = 0

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):

            if numbers[i] > numbers[j]:
                inversions += 1

    # For a 3x3 puzzle, even inversions = solvable
    return inversions % 2 == 0


# ------------------------------------------------------------
# 3. GENERATE POSSIBLE MOVES
# ------------------------------------------------------------

def get_neighbors(state):

    neighbors = []

    # Find blank position
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        # Check boundaries
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            # Create new state
            new_state = list(state)

            # Swap blank and tile
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(
                (tuple(new_state), move)
            )

    return neighbors


# ------------------------------------------------------------
# 4. DEPTH-LIMITED DFS
# ------------------------------------------------------------

def depth_limited_search(
    current,
    goal,
    depth_limit,
    path,
    moves,
    visited
):

    global nodes_explored

    nodes_explored += 1

    # Goal reached
    if current == goal:
        return True

    # Depth limit reached
    if depth_limit == 0:
        return False

    # Generate possible moves
    for next_state, move in get_neighbors(current):

        # Avoid states already present in current path
        if next_state not in visited:

            visited.add(next_state)

            path.append(next_state)
            moves.append(move)

            # Recursive DFS
            found = depth_limited_search(
                next_state,
                goal,
                depth_limit - 1,
                path,
                moves,
                visited
            )

            if found:
                return True

            # Backtrack
            path.pop()
            moves.pop()

            visited.remove(next_state)

    return False


# ------------------------------------------------------------
# 5. ITERATIVE DEEPENING SEARCH
# ------------------------------------------------------------

def iterative_deepening_search(
    start,
    goal,
    max_depth=31
):

    global nodes_explored

    total_nodes = 0

    # Try depth 0, 1, 2, 3, ...
    for depth in range(max_depth + 1):

        print("Searching at depth:", depth)

        nodes_explored = 0

        # Path contains states
        path = [start]

        # Moves contains moves
        moves = []

        # States in current DFS path
        visited = {start}

        # Perform depth-limited DFS
        found = depth_limited_search(
            start,
            goal,
            depth,
            path,
            moves,
            visited
        )

        total_nodes += nodes_explored

        print(
            "Nodes explored at this depth:",
            nodes_explored
        )

        if found:

            print("\nGoal found!")
            print(
                "Total nodes explored:",
                total_nodes
            )

            print(
                "Solution depth:",
                len(moves)
            )

            return path, moves

    print("\nNo solution found within depth", max_depth)

    print(
        "Total nodes explored:",
        total_nodes
    )

    return None, None


# ------------------------------------------------------------
# 6. INITIAL STATE
# ------------------------------------------------------------

initial_state = (
    5, 4, 0,
    6, 1, 8,
    7, 3, 2
)


# ------------------------------------------------------------
# 7. GOAL STATE
# ------------------------------------------------------------

goal_state = (
    0, 1, 2,
    3, 4, 5,
    6, 7, 8
)


# ------------------------------------------------------------
# 8. DISPLAY PUZZLE
# ------------------------------------------------------------

print("=" * 50)
print("8-PUZZLE USING ITERATIVE DEEPENING SEARCH")
print("=" * 50)

print("\nINITIAL STATE:")
print_board(initial_state)

print("GOAL STATE:")
print_board(goal_state)


# ------------------------------------------------------------
# 9. CHECK SOLVABILITY
# ------------------------------------------------------------

if not is_solvable(initial_state):

    print("No solution exists for this puzzle.")

else:

    print("Puzzle is solvable.\n")

    # --------------------------------------------------------
    # 10. RUN IDS
    # --------------------------------------------------------

    path, moves = iterative_deepening_search(
        initial_state,
        goal_state,
        max_depth=31
    )

    # --------------------------------------------------------
    # 11. DISPLAY SOLUTION
    # --------------------------------------------------------

    if path is not None:

        print("\n")
        print("=" * 50)
        print("SOLUTION PATH")
        print("=" * 50)

        print("\nTotal moves:", len(moves))

        print("\n")

        # Print initial state
        print("Step: 0")
        print("Initial State")
        print_board(path[0])

        # Print remaining states
        for i in range(1, len(path)):

            print("Step:", i)
            print("Move:", moves[i - 1])

            print_board(path[i])
```
