```python

# 4-Queens Problem using Hill-Climbing Search

def calculate_cost(state):
    """Count the number of pairs of queens attacking each other."""
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def get_neighbors(state):
    """Generate neighbors by swapping the rows of two queens."""
    neighbors = []
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            neighbor = state.copy()
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append((i, j, neighbor, calculate_cost(neighbor)))

    return neighbors


def hill_climbing(initial_state):
    current = initial_state.copy()
    step = 0

    while True:
        current_cost = calculate_cost(current)

        print(f"\nStep {step}")
        print("Current state:", current)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("Goal state reached!")
            break

        neighbors = get_neighbors(current)

        print("\nNeighbor states:")
        for i, j, state, cost in neighbors:
            print(f"Swap (x{i}, x{j}): {state}, Cost = {cost}")

        # Select the neighbor with the lowest cost.
        # Ties are resolved by the smallest column subscript pair.
        best_neighbor = min(neighbors, key=lambda x: (x[3], x[0], x[1]))

        if best_neighbor[3] >= current_cost:
            print("\nNo better neighbor found. Algorithm terminates.")
            break

        i, j, current, cost = best_neighbor
        print(f"\nSelected swap: (x{i}, x{j})")
        print("Next state:", current)
        print("Next cost:", cost)

        step += 1

    return current


# Initial state given in the question
initial_state = [3, 1, 2, 0]

print("Initial state:", initial_state)
solution = hill_climbing(initial_state)

print("\nFinal solution:", solution)
print("Final cost:", calculate_cost(solution))
```
