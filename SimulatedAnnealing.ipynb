```python

import random
import math

# Calculate number of attacking pairs
def cost(board):
    conflicts = 0

    for i in range(8):
        for j in range(i + 1, 8):

            # Same row or same diagonal
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Generate a random neighbour
def neighbour(board):
    new = board.copy()

    col = random.randint(0, 7)
    row = random.randint(0, 7)

    while row == new[col]:
        row = random.randint(0, 7)

    new[col] = row
    return new


# Simulated Annealing
def simulated_annealing():

    for restart in range(50):

        # Initial random state
        current = [random.randint(0, 7) for i in range(8)]
        T = 10

        while T > 0.1:

            # Generate next state
            next_state = neighbour(current)

            # Calculate cost difference
            deltaE = cost(current) - cost(next_state)

            if deltaE > 0:
                current = next_state

            else:
                p = math.exp(deltaE / T)

                if random.random() < p:
                    current = next_state

            # Check goal
            if cost(current) == 0:
                return current

            # Decrease temperature
            T *= 0.995

    return None


# Print solution
answer = simulated_annealing()

if answer is not None:
    print("Solution:", answer)

    for row in range(8):
        for col in range(8):
            if answer[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

else:
    print("Solution not found")
```
