```python

from itertools import product

print("A       B       C       A∨C     B∨¬C    KB      α")

for A, B, C in product([False, True], repeat=3):
    x = A or C
    y = B or (not C)
    KB = x and y
    alpha = A or B

    print(
        f"{str(A):7} {str(B):7} {str(C):7} "
        f"{str(x):7} {str(y):7} {str(KB):7} {alpha}"
    )

    if KB and not alpha:
        print("KB does not entail alpha")

print("\nKB entails alpha:",
      all(
          (A or B)
          for A, B, C in product([False, True], repeat=3)
          if ((A or C) and (B or (not C)))
      ))
```
