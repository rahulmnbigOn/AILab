```python

def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    # Apply existing substitutions
    x = substitute(x, subst)
    y = substitute(y, subst)

    # Step 1: If both are identical, return NIL
    if x == y:
        return subst

    # Step 2: If x is a variable
    if isinstance(x, str) and x.startswith('?'):
        if occurs(x, y):
            return None
        subst[x] = y
        return subst


    if isinstance(y, str) and y.startswith('?'):
        if occurs(y, x):
            return None
        subst[y] = x
        return subst


    if isinstance(x, (list, tuple)) and isinstance(y, (list, tuple)):


        if x[0] != y[0]:
            return None


        if len(x) != len(y):
            return None


        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst


    return None



def substitute(term, subst):
    if isinstance(term, str) and term.startswith('?'):
        while term in subst:
            term = subst[term]
        return term

    if isinstance(term, (list, tuple)):
        return type(term)(
            [substitute(t, subst) for t in term]
        )

    return term


def occurs(var, term):
    if var == term:
        return True

    if isinstance(term, (list, tuple)):
        return any(occurs(var, t) for t in term)

    return False



p1 = ('Knows', '?x', 'John')
p2 = ('Knows', 'Rahul', '?y')

result = unify(p1, p2)

if result is None:
    print("FAILURE")
elif result == {}:
    print("NIL")
else:
    print("Substitution:", result)
```
